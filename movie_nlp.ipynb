```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
def preprocess_text(text):

    text = str(text)
    text = text.lower()

    text = re.sub(r'[^a-zA-Z\s]', '', text)

    words = text.split()

    clean_words = []

    for word in words:
        if word not in stop_words:
            word = lemmatizer.lemmatize(word)
            clean_words.append(word)

    return " ".join(clean_words)
```

```python
df['tags'] = df['tags'].apply(preprocess_text)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    max_features=50000,
    ngram_range=(1, 2)
)

tfidf_matrix = tfidf.fit_transform(df['tags'])

tfidf_matrix
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def recommend(title, n=10):
    if title not in indices:
        return ['Movie not found']

    idx = indices[title]

    sim_score = cosine_similarity(
        tfidf_matrix[idx],
        tfidf_matrix
    ).flatten()

    similar_idx = sim_score.argsort()[::-1][1:n+1]

    return df['title'].iloc[similar_idx]
```

```python
recommend('Avatar', 5)
```

```python
def precision_at_5(movie, relevant_movies):
    recommendations = recommend(movie, 5)

    relevant_count = 0

    for recommendation in recommendations:
        if recommendation in relevant_movies:
            relevant_count += 1

    precision = relevant_count / 5

    return precision
```

```python
import numpy as np
import pandas as pd

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.metrics import ndcg_score
```

```python
df = df.reset_index(drop=True)

indices = pd.Series(
    df.index,
    index=df['title']
).drop_duplicates()

print("Number of movies:", len(df))
print("Number of unique titles:", df['title'].nunique())
```

```python
tfidf_uni = TfidfVectorizer(
    max_features=50000,
    ngram_range=(1, 1)
)

tfidf_uni_matrix = tfidf_uni.fit_transform(df['tags'])

print("Unigram TF-IDF shape:", tfidf_uni_matrix.shape)
```

```python
def recommend_unigram(title, n=10):

    if title not in indices:
        return []

    # Get the first index in case there are duplicate titles
    idx = indices[title]
    if isinstance(idx, pd.Series):
        idx = idx.iloc[0]

    similarity_scores = cosine_similarity(
        tfidf_uni_matrix[idx],
        tfidf_uni_matrix
    ).flatten()

    similar_indices = similarity_scores.argsort()[::-1][1:n+1]

    return df['title'].iloc[similar_indices].tolist()
```

```python
tfidf_bi = TfidfVectorizer(
    max_features=50000,
    ngram_range=(1, 2)
)

tfidf_bi_matrix = tfidf_bi.fit_transform(df['tags'])

print("Unigram + Bigram TF-IDF shape:", tfidf_bi_matrix.shape)
```

```python
def recommend_bigram(title, n=10):

    if title not in indices:
        return []

    # Get the first index in case there are duplicate titles
    idx = indices[title]
    if isinstance(idx, pd.Series):
        idx = idx.iloc[0]

    similarity_scores = cosine_similarity(
        tfidf_bi_matrix[idx],
        tfidf_bi_matrix
    ).flatten()

    similar_indices = similarity_scores.argsort()[::-1][1:n+1]

    return df['title'].iloc[similar_indices].tolist()
```

```python
def precision_at_k(recommendations, relevant_movies, k):

    recommendations = recommendations[:k]

    relevant_count = 0

    for movie in recommendations:
        if movie in relevant_movies:
            relevant_count += 1

    return relevant_count / k
```

```python
recommendations = [
    "Avatar 2",
    "The Inhabited Island",
    "Random Movie",
    "Moontrap: Target Earth",
    "Another Movie"
]

relevant = [
    "Avatar 2",
    "The Inhabited Island",
    "Moontrap: Target Earth"
]

print(precision_at_k(recommendations, relevant, 5))
```

```python
def recall_at_k(recommendations, relevant_movies, k):

    recommendations = recommendations[:k]

    relevant_count = 0

    for movie in recommendations:
        if movie in relevant_movies:
            relevant_count += 1

    if len(relevant_movies) == 0:
        return 0

    return relevant_count / len(relevant_movies)
```

```python
def ndcg_at_k(recommendations, relevant_movies, k):

    recommendations = recommendations[:k]

    relevance = []

    for movie in recommendations:
        if movie in relevant_movies:
            relevance.append(1)
        else:
            relevance.append(0)

    if len(relevance) == 0:
        return 0

    ideal_relevance = sorted(relevance, reverse=True)

    return ndcg_score(
        [ideal_relevance],
        [relevance]
    )
```

```python
recommendations = [
    "Avatar 2",
    "Random Movie",
    "The Inhabited Island",
    "Another Movie",
    "Moontrap: Target Earth"
]

relevant = [
    "Avatar 2",
    "The Inhabited Island",
    "Moontrap: Target Earth"
]

print(ndcg_at_k(recommendations, relevant, 5))
```

```python
def evaluate_model(recommend_function, test_movies):

    results = []

    for movie, relevant_movies in test_movies.items():

        recommendations = recommend_function(movie, 10)

        p5 = precision_at_k(
            recommendations,
            relevant_movies,
            5
        )

        p10 = precision_at_k(
            recommendations,
            relevant_movies,
            10
        )

        r5 = recall_at_k(
            recommendations,
            relevant_movies,
            5
        )

        r10 = recall_at_k(
            recommendations,
            relevant_movies,
            10
        )

        n5 = ndcg_at_k(
            recommendations,
            relevant_movies,
            5
        )

        n10 = ndcg_at_k(
            recommendations,
            relevant_movies,
            10
        )

        results.append({
            "Movie": movie,
            "Precision@5": p5,
            "Precision@10": p10,
            "Recall@5": r5,
            "Recall@10": r10,
            "NDCG@5": n5,
            "NDCG@10": n10
        })

    return pd.DataFrame(results)
```
